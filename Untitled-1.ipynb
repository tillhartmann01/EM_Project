```python
import numpy as np 
import math
from scipy.special import jv
import matplotlib.pyplot as plt
```

```python
f1 = 50 
f2 = 20000

d = np.linspace(0.1e-3, 0.5e-3, 100) #thicknes of the wire (in m)

w1 = 2*math.pi*f1
w2 = 2*math.pi*f2

L=1 #length of conductor


con_copper =  59.6e6 #conductivity of copper
my_0 = 4*math.pi*1e-7

skin_depth_1 = np.sqrt(2/(w1*1*my_0*con_copper))
skin_depth_2 = np.sqrt(2/(w2*1*my_0*con_copper))
print(skin_depth_1, skin_depth_2)

xi1 = d/(np.sqrt(2)*skin_depth_1)
xi2 = d/(np.sqrt(2)*skin_depth_2)
```

```python
def ber(n, xi):
    """Kelvin function ber of order n."""
    z = xi * np.exp(3j * np.pi / 4)
    return np.real(jv(n, z))


def bei(n, xi):
    """Kelvin function bei of order n."""
    z = xi * np.exp(3j * np.pi / 4)
    return np.imag(jv(n, z))


def G_R(xi, d):
    """
    Proximity-effect coefficient for a round conductor.

    Parameters
    ----------
    xi : float or ndarray
        Dimensionless parameter d / (sqrt(2) * delta).
    d : float
        Conductor diameter [m].

    Returns
    -------
    G_R : float or ndarray
        Proximity-effect coefficient [m^2].
    """

    ber0 = ber(0, xi)
    ber1 = ber(1, xi)
    ber2 = ber(2, xi)

    bei0 = bei(0, xi)
    bei1 = bei(1, xi)
    bei2 = bei(2, xi)

    denominator = ber0**2 + bei0**2

    term1 = (ber2 * ber1 + ber2 * bei1) / denominator

    term2 = (bei2 * bei1 - bei2 * ber1) / denominator

    GR = -(xi * np.pi**2 * d**2) / (2 * np.sqrt(2)) * (term1 + term2)

    return GR

def F_R(xi):
    """
    Skin-effect resistance factor for a round conductor.

    Parameters
    ----------
    xi : float or ndarray
        Dimensionless parameter d / (sqrt(2) * delta).

    Returns
    -------
    F_R : float or ndarray
        Skin-effect resistance factor (dimensionless).
    """

    ber0 = ber(0, xi)
    ber1 = ber(1, xi)

    bei0 = bei(0, xi)
    bei1 = bei(1, xi)

    denominator = ber1**2 + bei1**2

    term1 = (ber0 * bei1 - ber0 * ber1) / denominator

    term2 = (bei0 * ber1 + bei0 * bei1) / denominator

    FR = (xi / (4 * np.sqrt(2))) * (term1 - term2)

    return FR
```

```python
def r_dc(d_i, sigma=con_copper):
    """
    DC resistance per unit length of one strand [Ohm/m].
    """
    return 4 / (sigma * np.pi * d_i**2)


def P_dc_skin(xi, d_i, n_str, I_hat, L_coil):
    """
    DC + skin-effect losses of Litz wire [W].

    xi     : Dimensionless parameter
    d_i    : Individual strand diameter [m]
    n_str  : Number of strands
    I_hat  : Peak current through the complete Litz wire [A]
    L_coil : Total conductor length [m]
    """

    R = r_dc(d_i)

    P = n_str * R * F_R(xi) * (I_hat / n_str)**2 * L_coil

    return P


def P_internal_proximity(xi, d_i, d_a, n_str, I_hat, L_coil):
    """
    Internal proximity-effect losses of Litz wire [W].

    xi     : Dimensionless parameter
    d_i    : Individual strand diameter [m]
    d_a    : Outer diameter of complete Litz wire [m]
    n_str  : Number of strands
    I_hat  : Peak current through the complete Litz wire [A]
    L_coil : Total conductor length [m]
    """

    R = r_dc(d_i)

    P = (n_str * R * G_R(xi, d_i)
         * I_hat**2 / (2 * np.pi**2 * d_a**2)
         * L_coil)

    return P


def P_litz_total(xi, d_i, d_a, n_str, I_hat, L_coil):
    """
    Total losses from DC, skin effect and internal proximity effect [W].
    """

    P1 = P_dc_skin(xi, d_i, n_str, I_hat, L_coil)

    P2 = P_internal_proximity(xi, d_i, d_a, n_str, I_hat, L_coil)

    return P1 + P2
```

```python
# Parameters for a single insulated copper wire
n_str = 1
d_i = d
d_a = 1.2 * d          # 10% insulation thickness on each side
L_coil = L             # 1 meter
I_hat = 1              # Peak current [A]

# Calculate internal proximity losses
P_prox_50 = P_internal_proximity(
    xi1, d_i, d_a, n_str, I_hat, L_coil
)

P_prox_20k = P_internal_proximity(
    xi2, d_i, d_a, n_str, I_hat, L_coil
)

# Plot
plt.figure(figsize=(9, 5))

plt.plot(d * 1e3, P_prox_50, label="50 Hz")
plt.plot(d * 1e3, P_prox_20k, label="20 kHz")

plt.xlabel("Copper wire diameter [mm]")
plt.ylabel("Internal proximity losses [W]")
plt.title("Internal Proximity Losses vs. Wire Diameter")

plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import re

def load_femm_H_contour(filename):
    """
    Load magnetic field data exported from FEMM.

    Assumed columns:
        Column 1: Position [mm]
        Column 2: |H| [A/m]

    Returns
    -------
    x : ndarray
        Position along FEMM contour [mm]
    H : ndarray
        Magnetic field strength [A/m]
    """

    data = []

    with open(filename, "r", encoding="utf-8-sig", errors="replace") as file:
        for line in file:

            line = line.strip()

            # Skip empty lines and comments
            if not line or line.startswith(("#", "%", "//")):
                continue

            # Handle tabs, spaces, semicolons, and commas
            values = re.split(r'[\s;,]+', line)

            if len(values) < 2:
                continue

            try:
                # Support scientific notation, including Fortran D notation
                x = float(values[0].replace("D", "E").replace("d", "e"))
                H = float(values[1].replace("D", "E").replace("d", "e"))

                if np.isfinite(x) and np.isfinite(H):
                    data.append([x, H])

            except ValueError:
                continue

    if len(data) < 2:
        raise ValueError(
            "No valid numeric data found in FEMM export.\n"
            "Check that H_contour.txt contains two numeric columns."
        )

    data = np.array(data)

    return data[:, 0], data[:, 1]


# ==========================================
# Import FEMM data
# ==========================================

x_femm, H_femm = load_femm_H_contour("H_contour.txt")


# ==========================================
# Plot imported magnetic field
# ==========================================

plt.figure(figsize=(9, 5))

plt.plot(x_femm, H_femm, 'k-', linewidth=2)

plt.xlabel("Position [mm]")
plt.ylabel("|H| [A/m]")
plt.title("Magnetic Field from FEMM")

plt.grid(True)
plt.tight_layout()
plt.show()


# ==========================================
# Calculate integral of |H|² dl
# ==========================================

# Convert position from mm to meters
x_m = x_femm * 1e-3

# Numerical integration using trapezoidal rule
I_H2 = np.trapezoid(H_femm**2, x_m)

print("===================================")
print("FEMM Magnetic Field Integral")
print("===================================")
print(f"Maximum |H| = {np.max(H_femm):.4e} A/m")
print(f"Integral of |H|² dl = {I_H2:.6e} A²/m")

"""
# ==========================================
# Plot |H|² along contour
# ==========================================

plt.figure(figsize=(9, 5))

plt.plot(x_femm, H_femm**2, 'b-', linewidth=2)

plt.xlabel("Position [mm]")
plt.ylabel("|H|² [A²/m²]")
plt.title("Squared Magnetic Field Along FEMM Contour")

plt.grid(True)
plt.tight_layout()
plt.show()
"""
```

```python
def P_external_proximity(xi, d_i, H_ext, L_turns, n_str=1):
    """
    External proximity-effect losses [W].

    Parameters
    ----------
    xi : float or ndarray
        Dimensionless skin-depth parameter.

    d_i : float or ndarray
        Copper strand diameter [m].

    H_ext : float or ndarray
        Peak external magnetic field strength at each wire [A/m].

    L_turns : float or ndarray
        Length of each turn or wire segment [m].

    n_str : int
        Number of strands (1 for a solid wire).

    Returns
    -------
    P_ex : float or ndarray
        External proximity-effect losses [W].
    """

    # DC resistance per unit length
    R_dc = 4 / (con_copper * np.pi * d_i**2)

    # Proximity-effect coefficient
    GR = G_R(xi, d_i)

    # Field strength and conductor lengths
    H_ext = np.atleast_1d(np.asarray(H_ext, dtype=float))
    L_turns = np.broadcast_to(
        np.asarray(L_turns, dtype=float),
        H_ext.shape
    )

    # Sum contributions from all turns
    H_integral = np.sum(H_ext**2 * L_turns)

    # External proximity losses
    P_ex = n_str * R_dc * GR * H_integral

    return P_ex

# Single copper conductor
n_str = 1

# Position of wire along FEMM contour [mm]
# Example only: replace with actual wire-centre position
x_wire = 12.0

# Interpolate magnetic field at wire position
H_ext = np.interp(x_wire, x_femm, H_femm)

print(f"External magnetic field: {H_ext:.2f} A/m")


# Calculate external proximity losses at 50 Hz
P_ex_50 = P_external_proximity(
    xi1, d, H_ext, L, n_str
)

# Calculate external proximity losses at 20 kHz
P_ex_20k = P_external_proximity(
    xi2, d, H_ext, L, n_str
)

plt.figure(figsize=(9, 5))

plt.plot(d * 1e3, P_ex_50, label="50 Hz")
plt.plot(d * 1e3, P_ex_20k, label="20 kHz")

plt.xlabel("Copper wire diameter [mm]")
plt.ylabel("External proximity losses [W]")
plt.title("External Proximity Losses vs. Wire Diameter")

plt.legend()
plt.grid(True)
plt.show()
```
